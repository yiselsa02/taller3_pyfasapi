### 3. Aplique normalizando y estandarizacion.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler

# Cargar el dataset
df = pd.read_excel("dataset_ventas.xlsx")

# Variables numéricas
columnas_numericas = ['cantidad', 'precio_unitario', 'total_venta']

# Normalización
normalizador = MinMaxScaler()

df_normalizado = df.copy()

df_normalizado[columnas_numericas] = normalizador.fit_transform(
    df[columnas_numericas]
)

print("DATOS NORMALIZADOS:")
print(df_normalizado.head())

# Estandarización
estandarizador = StandardScaler()

df_estandarizado = df.copy()

df_estandarizado[columnas_numericas] = estandarizador.fit_transform(
    df[columnas_numericas]
)

print("\nDATOS ESTANDARIZADOS:")
print(df_estandarizado.head())
```

### 4. Balancee la variable objetivo usando SMOTE

```python
from imblearn.over_sampling import SMOTE
import pandas as pd

# Cargar dataset
df = pd.read_excel("dataset_ventas.xlsx")

# Convertir fecha a números
df['fecha'] = pd.to_datetime(df['fecha'])
df['año'] = df['fecha'].dt.year
df['mes'] = df['fecha'].dt.month
df['día'] = df['fecha'].dt.day

# Eliminar la fecha original
df = df.drop('fecha', axis=1)

# Convertir variables categóricas a variables numéricas
df = pd.get_dummies(
    df,
    columns=['ciudad', 'producto'],
    dtype=int
)

# Separar variables predictoras y variable objetivo
X = df.drop('categoria', axis=1)
y = df['categoria']

# Convertir la variable objetivo a números
y = y.astype('category').cat.codes

# Aplicar SMOTE
smote = SMOTE(random_state=42)

X_balanceado, y_balanceado = smote.fit_resample(X, y)

# Verificar el balance antes y después
print("Antes del balanceo:")
print(y.value_counts())

print("\nDespués del balanceo:")
print(y_balanceado.value_counts())
```

### 5. Muestre el conjunto de datos final listo para entrenar un modelo de clasificacion.

```python
# Crear el conjunto de datos final
df_final = X_balanceado.copy()
df_final['categoria'] = y_balanceado

# Mostrar el conjunto de datos final
print("Conjunto de datos final listo para entrenar el modelo:")
display(df_final)

# Mostrar dimensiones
print("\nDimensiones del conjunto final:")
print(df_final.shape)

# Verificar que no haya valores nulos
print("\nValores nulos:")
print(df_final.isnull().sum())
```
