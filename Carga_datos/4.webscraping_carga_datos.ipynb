```python
import pandas as pd
import requests
from io import StringIO

# URL de Wikipedia
url = "https://es.wikipedia.org/wiki/Anexo:Pa%C3%ADses_y_territorios_dependientes_por_poblaci%C3%B3n"

# Encabezados
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 Chrome/139.0.0.0 Safari/537.36"
}

# Hacer la petición
respuesta = requests.get(url, headers=headers)

# Verificar que funcionó
respuesta.raise_for_status()

print("Página descargada correctamente")

# Convertir el HTML en texto para pandas
html = StringIO(respuesta.text)

# Buscar las tablas
tables = pd.read_html(html)

print("Cantidad de tablas encontradas:", len(tables))

# Mostrar las tablas encontradas
for i, tabla in enumerate(tables):
    print(f"\n--- TABLA {i} ---")
    print(tabla.head())

# Seleccionar la primera tabla
df_web = tables[0]

print("\nPrimera tabla:")
print(df_web.head())
```
